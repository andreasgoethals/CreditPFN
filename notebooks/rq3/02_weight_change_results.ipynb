# RQ3: controlling weight change — results

Five-fold evaluation of update size × anchoring (E8) and trainable scope (E9) against the original models: the transfer–forgetting trade-off across all arms (one figure per task), with the per-arm and per-table numbers in tables. Unit of analysis: one value per table; arm summaries are equal-table means over tables with all five paired folds.

```python
%matplotlib inline
import sys
from pathlib import Path
REPO = next(p for p in (Path.cwd(), *Path.cwd().parents) if (p / 'pyproject.toml').is_file())
if str(REPO) not in sys.path:
    sys.path.insert(0, str(REPO))
import pandas as pd
from src.visualize import style, campaign as cp, campaign2 as c2, results2 as r2
from src.visualize.figures import FigureSaver
style.apply()
sink = FigureSaver('rq3/02_weight_change_results')
report = cp.NotebookReport('RQ3: controlling weight change — results', sink=sink)
overview = {track: c2.arms_overview(['reference', 'update_size_anchoring', 'scope'], track) for track in ('pd', 'lgd')}
```

## 1. Learning rate and L2-SP (E8)

Question: how does the five-fold effect on held-out credit tables and on the retention panel respond to the peak learning rate, and does anchoring (L2-SP) change that response? The tables below give the numbers; section 2 draws them, with the learning rate as marker size and L2-SP strength as marker shape.

```python
for track in ('pd', 'lgd'):
    gains = c2.endpoint_gains('update_size_anchoring', track)
    report.table(track.upper() + ' E8 summary per arm and model', c2.gain_summary(gains))
    report.table(track.upper() + ' E8 per-table gains', gains)
report.add('1. Learning rate and L2-SP (E8)', c2.evidence_note('E8 five-fold gains', c2.select(overview['pd'], slug='update_size_anchoring')))
```

## 2. Transfer against forgetting across recipes (E1, E8, E9)

Question: which recipes gain on held-out credit tables at the smallest loss on non-credit tables, and does a restricted trainable scope buy a better trade-off than a smaller update?

```python
for track in ('pd', 'lgd'):
    fig = r2.tradeoff(overview[track], track=track, title='Transfer against forgetting per recipe (E1, E8, E9)')
    cp.show(sink, c2.pages(fig, 'tradeoff',
        'Equal-table mean five-fold change on the non-credit retention panel (vertical) against that on the held-out '
        'credit tables (horizontal) for every arm of E1, E8 and E9, per base model (columns). Colour: trainable scope '
        '(black: all weights); circle, downward and upward triangle: all weights with L2-SP λ = 0.003, 0 and 0.03; squares: '
        'frozen backbone, DiagScale and LoRA (λ = 0.003). Marker area grows with the peak learning rate (the legend gives the smallest and largest), and lines join '
        'the learning rates of one scope and strength. ' + r2.SCORE[track] + '; tables with all five paired folds only.'), track=track)
    report.table(track.upper() + ' arm overview (credit and retention means)', overview[track])
report.add('2. Transfer against forgetting across recipes (E1, E8, E9)', c2.evidence_note('E8 and E9 arm overview', overview['pd']))
```

## 3. Trainable scope in detail (E9) and calibration per arm

Per-table five-fold gains of every E9 arm, and the metric profile (discrimination, calibration, distribution quality) of every E8 and E9 arm on held-out credit tables.

```python
for track in ('pd', 'lgd'):
    gains = c2.endpoint_gains('scope', track)
    report.table(track.upper() + ' E9 summary per arm and model', c2.gain_summary(gains))
    report.table(track.upper() + ' E9 per-table gains', gains)
    for slug in ('update_size_anchoring', 'scope'):
        report.table(track.upper() + ' ' + slug + ' metric profile per arm', c2.metric_profile(c2.endpoint_metrics(slug, track)))
report.add('3. Trainable scope in detail (E9) and calibration per arm', 'Tables only; the figures above summarise the same gains.')
```
