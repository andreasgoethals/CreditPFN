# Appendix, public tables: RQ3: training data — results

**Appendix: public tables only.** Every section below is the one of `notebooks/rq3/04_training_data_results.ipynb`, recomputed after removing the proprietary credit tables from the evaluation (15 of 17 PD and 3 of 8 LGD held-out tables remain). The CPT models are the same: their training corpora still contained the proprietary tables, so this appendix checks the evaluation on data a reader can obtain, not the training. With 3 LGD tables, LGD means rest on very few tables; the per-table rows are the evidence. In the E7 transfer matrices, proprietary tables still appear as sources (they were trained on) but no longer as targets.

Five-fold evaluation of the training-data arms against the original models: full pass (E10), small versus count-matched large tables (E11) and episode size (E12, also at matched inference context), and what each costs. Unit of analysis: one value per table, folds averaged first, complete five-fold tables only. E1 is the control of E10; the two E11 corpora are compared with each other; E12's arms share the corpus of tables above 10,000 rows.

```python
%matplotlib inline
import sys
from pathlib import Path
REPO = next(p for p in (Path.cwd(), *Path.cwd().parents) if (p / 'pyproject.toml').is_file())
if str(REPO) not in sys.path:
    sys.path.insert(0, str(REPO))
import pandas as pd
from src.visualize import style, campaign as cp, campaign2 as c2, results2 as r2
from src.visualize.figures import FigureSaver
style.apply()
c2.use_tables('public')
sink = FigureSaver('appendix/05_rq3_training_data_results')
report = cp.NotebookReport('Appendix, public tables: RQ3: training data — results', sink=sink)
slugs = ['reference', 'exposure', 'table_size', 'episode_size']
order = ['Reference', 'Full pass', '10,000 updates', 'Small tables only', 'Large tables, count-matched', 'Episodes of 2,048 rows', 'Episodes of 4,096 rows', 'Episodes of 8,192 rows', 'Episodes of 16,384 rows', 'Reference episode size']
```

```python
report.table('Held-out credit tables in scope', pd.concat([c2.scope_counts(t) for t in ('pd', 'lgd')], ignore_index=True))
```

## 1. Effect of each training-data arm (E10–E12)

Question: does any change to the training data move the five-fold effect on held-out credit tables, or on the retention panel, beyond the reference?

```python
for track in ('pd', 'lgd'):
    table = c2.arm_table_gains(slugs, track)
    fig = r2.arm_gains(table, order, track=track, title='Five-fold effect of each training-data arm')
    cp.show(sink, c2.pages(fig, 'data_arms',
        'Five-fold change of ' + r2.SCORE[track] + ' (CPT minus the original model on identical rows) per training-data '
        'arm (rows) and base model (colour): equal-table mean (dot) and interquartile range over tables (line), on the '
        'held-out credit tables (left) and the non-credit retention panel (right). Arms: reference (E1), '
        'full pass (E10), small tables only and the same number of large tables (E11, PD), and episodes '
        'of 2,048 to 16,384 rows on the tables above 10,000 rows (E12, PD). Tables with all five paired folds only.'), track=track)
    for slug in slugs[1:]:
        gains = c2.endpoint_gains(slug, track)
        report.table(track.upper() + ' ' + slug + ' summary per arm and model', c2.gain_summary(gains))
        report.table(track.upper() + ' ' + slug + ' per-table gains', gains)
report.add('1. Effect of each training-data arm (E10-E12)', c2.evidence_note('E10-E12 five-fold gains', c2.arm_table_gains(slugs[1:], 'pd')))
```

## 2. Episode size at matched inference context (E12)

Question: is a model trained on larger episodes better at every inference context size, or only when it is given a large context? Each E12 model is scored with 2,048 and 8,192 labelled context rows and the full pool, minus the original model scored by E6 on the same rows.

```python
matched = c2.episode_context_gains(c2.task_results('episode_context'), c2.task_results('context'))
fig = r2.episode_matched_context(matched, title='Episode size at matched inference context (E12)')
cp.show(sink, c2.pages(fig, 'episode_matched_context',
    'Equal-table mean five-fold AUC change over the original model at the same inference context (colour: 2,048 and '
    '8,192 labelled rows, and the full pool capped at the model\'s limit) against the training-episode size of E12, per '
    'base model (columns; k = 1,024 rows; the last, unlabelled point is the reference episode size of 10,000 or 26,000 rows). The original model\'s scores come from the E6 context task on the same folds and context '
    'rows; tables with all five folds at that context only.'), track='pd')
report.table('E12 matched-context gains per table', matched)
report.table('E12 AUC per arm, model and context size', c2.episode_context_summary(c2.task_results('episode_context')))
report.add('2. Episode size at matched inference context (E12)', c2.evidence_note('E12 matched-context evaluation', matched))
```

## 3. Effect against training cost

Question: what does each exposure or episode-size arm cost per training run, and does the cost buy a larger effect? A table: effect and wall-clock hours per arm and model.

```python
overview = pd.concat([c2.arms_overview(['reference', 'exposure', 'episode_size'], track) for track in ('pd', 'lgd')], ignore_index=True)
resources = c2.resource_summary(c2.run_resources(c2.load_trajectories(['reference', 'exposure', 'episode_size'])))
report.table('Effect and cost per arm and model', c2.join(overview, resources, on=['slug', 'arm', 'track', 'model']))
report.add('3. Effect against training cost', c2.evidence_note('cost records', resources))
```
