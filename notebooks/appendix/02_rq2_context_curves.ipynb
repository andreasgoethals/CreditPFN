# Appendix, public tables: RQ2: context scarcity (E6)

**Appendix: public tables only.** Every section below is the one of `notebooks/rq2/01_context_curves.ipynb`, recomputed after removing the proprietary credit tables from the evaluation (15 of 17 PD and 3 of 8 LGD held-out tables remain). The CPT models are the same: their training corpora still contained the proprietary tables, so this appendix checks the evaluation on data a reader can obtain, not the training. With 3 LGD tables, LGD means rest on very few tables; the per-table rows are the evidence. In the E7 transfer matrices, proprietary tables still appear as sources (they were trained on) but no longer as targets.

Question: is CPT worth more when the target table offers few labelled context rows, and is the part that helps credit-specific? The original, the E1 model (credit CPT) and the E4 model (CPT on the matched non-credit twins) are scored with nested labelled contexts from 8 rows to the full pool on all five folds, with identical validation and test rows (the fold's validation rows stay labelled for calibration and thresholds, so this is context scarcity rather than total label scarcity).

Larger contexts exist only for larger tables. Equal-table means are therefore drawn only up to the largest size that every table of the model reached (a fixed cohort); per-table curves continue to each table's own largest grid size, and full-pool scores are tabulated.

```python
%matplotlib inline
import sys
from pathlib import Path
REPO = next(p for p in (Path.cwd(), *Path.cwd().parents) if (p / 'pyproject.toml').is_file())
if str(REPO) not in sys.path:
    sys.path.insert(0, str(REPO))
import pandas as pd
from src.visualize import style, campaign as cp, campaign2 as c2, results2 as r2
from src.visualize.figures import FigureSaver
style.apply()
c2.use_tables('public')
sink = FigureSaver('appendix/02_rq2_context_curves')
report = cp.NotebookReport('Appendix, public tables: RQ2: context scarcity (E6)', sink=sink)
results = c2.task_results('context')
table = c2.context_table(results)
curve = c2.context_cohort_curve(table)
```

```python
report.table('Held-out credit tables in scope', pd.concat([c2.scope_counts(t) for t in ('pd', 'lgd')], ignore_index=True))
```

## 1. Context curves

The gain of the CPT model over the original at each number of labelled context rows, per table and as the fixed-cohort mean, with the fixed-cohort mean of the non-credit CPT model (E4) on the same rows. Absolute scores are in the tables.

```python
fig = r2.context_curves(table, curve, title='CPT gain against labelled context rows (E6)')
cp.show(sink, c2.pages(fig, 'context_curves',
    'Five-fold change of the E1 CPT model over the original model against the number of labelled context rows (log '
    'scale), per task (rows; AUC for PD, R² for LGD) and base model (columns; TabPFN v2 is capped at 10,000 rows). '
    'Grey: one held-out credit table, up to its largest grid size. Coloured: equal-table mean over the tables that reach '
    'every plotted size (fixed cohort); dotted: E4 on the same cohort, shown only where every contributing table has all five E4 folds. Context prefixes, validation and test rows are identical for both models; '
    'tables with all five folds at a size only.'))
for track in ('pd', 'lgd'):
    report.table(track.upper() + ' equal-table curve per size (fixed_cohort marks the plotted range)', c2.select(curve, track=track))
    report.table(track.upper() + ' per table and size', c2.select(table, track=track))
report.add('1. Context curves', c2.evidence_note('E6 context curves', table))
```

## 2. How much does the gain change when labelled context doubles?

Fit a slope to each held-out table over the common context sizes reached by every table of that model. Each estimate needs at least three sizes and complete observations for the contrast. A negative slope means the adaptation gain shrinks as more labels become available. The table gives the equal-table mean slope and its 95% interval from resampling tables.

```python
slopes = c2.context_slopes(table)
slope_summary = c2.slope_summary(slopes)
report.table('Slope per table and quantity', slopes)
report.table('Mean slope per task, model and quantity', slope_summary)
report.add('2. Slope test', c2.evidence_note('E6 slope test', slopes))
```
